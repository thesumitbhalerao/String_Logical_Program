##### 1) Reverse a String :-
##### - I reverse the given string and display it.

```python
s="sumit"
reverse=s[::-1]
print("Original String :",s)
print("Reverse String :",reverse)
```

#### 2) Palindrome String :-
##### - first reverse the string then compare reverse and original string if same then palindrome if not then no palindrome

```python
s="sumit"
if s==s[::-1]:
    print("String is palindrome")
else:
    print("String is not palindrome")

y="nayan"
if y==y[::-1]:
    print("String is palindrome")
else:
    print("String is not palindrome")
```

#### 3) Count Vowels in string :-
##### - I count how many vowels are present in the string
##### - Check every character, if character is vowels then increase the count.

```python
a="The QUick BrOwn FOx Jumps OvEr tHe lAzy dOg"
count=0
for ch in a.lower():
    if ch in "aeiou":
        count+=1
print("Number of vowels in the string:",count)
```

#### 4) Count Consonants in string :-
##### - I count how many consonants are present in string
##### - Checks every character, if character is consonants then increase the count.

```python
a="The QUick BrOwn FOx Jumps OvEr tHe lAzy dOg"
count=0
for ch in a.lower():
    if ch.isalpha() and ch not in "aeiou":
        count+=1
print("Number of consonants in the string:",count)
```

#### 5) Count both vowels and consonants

```python
s="Sumit Bhalero"
vowels=0
consonants=0
for ch in s.lower():
    if ch in "aeiou":
        vowels+=1
    elif ch.isalpha():
        consonants+=1
print("Number of vowels in the string:",vowels)
print("Number of consonants in the string:",consonants)
```

#### 6) Counts Digits in a String
##### - I count how many numeric characters are present

```python
s="Sumit1234Bhalerao"
count=0
for ch in s:
    if ch.isdigit():
        count+=1
print("Number of digits in the string:",count)
```

#### 7) Count Spaces in string
##### I count the space are in string

```python
s="Sumit Bhalerao   "
count=0
for ch in s:
    if ch==" ":
        count+=1
print("Number of spaces in the string:",count)
```

#### 8) Check String starts and ends with same character


```python
s="My name is Shyam"
if s[0].lower()==s[-1].lower():
    print("First and last character are same")
else:
    print("First and last character are not same")
```

#### 9) Find Missing Alphabet
##### To find missing alphabet from given string


```python
a=("The Kiran Academy")
missing=""
count=0
for ch in "abcdefghijklmnoqprstuvwxyz":
    if ch not in a.lower():
        missing+=ch
        count+=1
print("missing alphabet:",missing)
print("counting missing alphabet:",count)
```

#### 10) Count Alphabets, Digits and Special Character

```python
s="sumitbhalerao20#03@gmail.com"
alpha=0
digit=0
special=0
for ch in s:
    if ch.isalpha():
        alpha+=1
    elif ch.isdigit():
        digit+=1
    else:
        special+=1
print("Alphabets :",alpha)
print("Digits :",digit)
print("Special :",special)
```
